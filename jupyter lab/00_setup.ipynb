## Shared files
Paste the two Google Drive files. The ball model must be the trained volleyball ball model.

```python
import gdown
BALL_MODEL_URL = ""  # Paste the shared ball model link here.
SAMPLE_VIDEO_URL = ""  # Paste the shared video link here.
for url, target in [(BALL_MODEL_URL, "models/ball_yolov8n.pt"),
                    (SAMPLE_VIDEO_URL, "input_videos/volleyball_sample.mp4")]:
    if not url:
        print("Missing shared link for:", target)
    elif not Path(target).exists():
        result = gdown.download(url, target, fuzzy=True, quiet=False)
        if result is None:
            raise RuntimeError("Download failed: " + target)
```
